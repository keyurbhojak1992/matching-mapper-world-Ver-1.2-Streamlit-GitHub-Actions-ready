## Step 3 — Upload actual production inputs

Upload one ZIP containing Sales/BID, Actions, and current Stock. The validated control workbook is already embedded, so no control upload is required for normal use.


```python
from google.colab import files
from pathlib import Path
import zipfile, shutil
UPLOAD_DIR=Path('/content/dealmaker_uploads_v8_2')
if UPLOAD_DIR.exists(): shutil.rmtree(UPLOAD_DIR)
UPLOAD_DIR.mkdir(parents=True)
USE_INPUT_ZIP=True
UPLOAD_CUSTOM_CONTROL=False  # Embedded validated control is the default

def upload_one(label):
    print(f'Upload exactly one file for: {label}')
    uploaded=files.upload()
    if len(uploaded)!=1: raise ValueError(f'Expected exactly one {label} file, received {len(uploaded)} files.')
    name=next(iter(uploaded)); path=UPLOAD_DIR/name; path.write_bytes(uploaded[name]); return path

if USE_INPUT_ZIP:
    zpath=upload_one('INPUT ZIP')
    extract=UPLOAD_DIR/'input'; extract.mkdir()
    with zipfile.ZipFile(zpath) as zf: zf.extractall(extract)
    books=[p for p in extract.rglob('*') if p.suffix.lower() in {'.xlsx','.xlsm','.csv','.tsv'} and not p.name.startswith('~$')]
    def find_one(words):
        hits=[p for p in books if any(w in p.name.lower() for w in words)]
        if len(hits)!=1: raise ValueError(f'Could not uniquely identify {words}. Found: {[p.name for p in hits]}')
        return hits[0]
    PROD_SALES=find_one(['sales','sale','bid'])
    PROD_ACTIONS=find_one(['action'])
    PROD_STOCK=find_one(['stock','inventory'])
else:
    PROD_SALES=upload_one('FULL SALES/BID HISTORY')
    PROD_ACTIONS=upload_one('FULL ACTIONS HISTORY')
    PROD_STOCK=upload_one('ACTUAL CURRENT STOCK')
PROD_CONTROL=upload_one('CUSTOM CONTROL WORKBOOK') if UPLOAD_CUSTOM_CONTROL else DEFAULT_CONTROL
SALES_SHEET=ACTIONS_SHEET=STOCK_SHEET=''
SALES_HEADER_ROW_EXCEL=ACTIONS_HEADER_ROW_EXCEL=STOCK_HEADER_ROW_EXCEL=None
print('Sales:',PROD_SALES.name,'Actions:',PROD_ACTIONS.name,'Stock:',PROD_STOCK.name,'Control:',PROD_CONTROL.name)
```

## Step 4 — Run the v8.2.2 preflight


```python
import subprocess, sys, pandas as pd
PREFLIGHT_OUTPUT=PROJECT_DIR/'Dealmaker_Data_Preflight_v6.xlsx'

def add_optional(cmd,flag,value):
    if value not in (None,''):
        cmd.extend([flag,str(value)])

def zero_based(row):
    return None if row in (None,'') else int(row)-1

def run_visible(cmd,label):
    print(f'\n--- {label} ---')
    print(' '.join(map(str,cmd)))
    process=subprocess.Popen(cmd,stdout=subprocess.PIPE,stderr=subprocess.STDOUT,text=True,bufsize=1)
    tail=[]
    assert process.stdout is not None
    for line in process.stdout:
        print(line,end='')
        tail.append(line.rstrip())
        if len(tail)>100:
            tail.pop(0)
    code=process.wait()
    if code!=0:
        raise RuntimeError(f'{label} failed with exit code {code}. Last output:\n'+'\n'.join(tail))

cmd=[sys.executable,str(PREFLIGHT),'--sales',str(PROD_SALES),'--actions',str(PROD_ACTIONS),'--stock',str(PROD_STOCK),'--control',str(PROD_CONTROL),'--output',str(PREFLIGHT_OUTPUT)]
for flag,val in [('--sales-sheet',SALES_SHEET),('--actions-sheet',ACTIONS_SHEET),('--stock-sheet',STOCK_SHEET),('--sales-header-row',zero_based(SALES_HEADER_ROW_EXCEL)),('--actions-header-row',zero_based(ACTIONS_HEADER_ROW_EXCEL)),('--stock-header-row',zero_based(STOCK_HEADER_ROW_EXCEL))]:
    add_optional(cmd,flag,val)
run_visible(cmd,'Preflight')
print(PREFLIGHT_OUTPUT)
```

## Step 5 — Review preflight and block on errors


```python
issues=pd.read_excel(PREFLIGHT_OUTPUT,sheet_name='Issues')
display(issues)
BLOCKING_ERRORS=int(issues['Severity'].astype(str).str.upper().eq('ERROR').sum())
print('Blocking errors:',BLOCKING_ERRORS)
if BLOCKING_ERRORS:
    files.download(str(PREFLIGHT_OUTPUT))
    raise RuntimeError('Correct all ERROR rows before production. Review Quality_Domain_Audit first.')
```

## Step 6 — Run the memory-safe v8.2.2 production generator and independent audit


```python
PROD_OUTPUT=PROJECT_DIR/'production_output_v8_2_2'
cmd=[sys.executable,str(RUNNER),'--sales',str(PROD_SALES),'--actions',str(PROD_ACTIONS),'--stock',str(PROD_STOCK),'--control',str(PROD_CONTROL),'--output-dir',str(PROD_OUTPUT),'--chunk-size','10']
for flag,val in [('--sales-sheet',SALES_SHEET),('--actions-sheet',ACTIONS_SHEET),('--stock-sheet',STOCK_SHEET),('--sales-header-row',zero_based(SALES_HEADER_ROW_EXCEL)),('--actions-header-row',zero_based(ACTIONS_HEADER_ROW_EXCEL)),('--stock-header-row',zero_based(STOCK_HEADER_ROW_EXCEL))]:
    add_optional(cmd,flag,val)
run_visible(cmd,'Production')
print('Output:',PROD_OUTPUT)
```
